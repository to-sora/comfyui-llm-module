```python
import subprocess, pathlib, os
ROOT = pathlib.Path("/content/comfyui-llm-module") if pathlib.Path("/content").exists() else pathlib.Path.cwd()/"comfyui-llm-module"
if not ROOT.exists():
    subprocess.run(["git", "clone", "-4", "--branch", "feat/qwen-comfy-colab", "https://github.com/to-sora/comfyui-llm-module.git", str(ROOT)], check=True)
APP = ROOT/"plugin"
subprocess.run(["bash", str(APP/"install-app-dep.sh")], check=True)
PY = str(APP/".local-tool-app/venv/bin/python")
```

```python
MODEL = "Qwen/Qwen3.5-0.8B"
subprocess.run([PY, str(APP/"src/acceptance/prepare.py"), MODEL], check=True)
subprocess.run(["bash", str(APP/"start.sh")], check=True)
```

```python
from IPython.display import display, HTML
try:
    result = subprocess.run([PY, str(APP/"src/acceptance/run.py")])
    report = ROOT/"fan-out/acceptance.html"
    if report.exists():
        display(HTML(report.read_text()))
    result.check_returncode()
finally:
    subprocess.run(["bash", str(APP/"start.sh"), "--stop"], check=True)
```
